```python
import numpy as np
```

```python
#operations on array..
arr = np.array([[1,2,3],[4,5,6]])
print(arr,arr.shape)
reshaped = arr.reshape((3,2))
print(reshaped,reshaped.shape)

flattened = arr.flatten() # convert 2D =>array to 1D array..
print(flattened,flattened.shape)
```

```python
# Indexing 
arr = np.array([1,2,3,4,5])
print(arr[0])
print(arr[3])
```

```python
# for 2D array Indexing..
arrd = np.array([[1,2,3],[4,5,6]])
print(arrd[0][1])
```

```python
# Fancy Indexing
arr = np.array([1,2,3,4,5])
idx = [0,1,3]
print(arr[idx])
```

```python
#Boolean Indexing
arr = np.array([1,2,3,4,5])
print(arr[arr>2])
print(arr[arr%2==0])
print(arr[arr%2!=0])
```

```python
# Slicing..[start,end,step]
arr = np.array([1,2,3,4,5])
print(arr[1:4])
print(arr[1:])
print(arr[:4])
```
