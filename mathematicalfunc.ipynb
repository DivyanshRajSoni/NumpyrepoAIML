```python
#mathematical Functions.
#Aggregation
import numpy as np
arr =np.array([1,2,3,4,5])
print(np.sum(arr))
print(np.prod(arr))
print(np.min(arr))
print(np.max(arr))
print(np.argmin(arr)) # to return the index of min value
print(np.argmax(arr)) # to return the index of the max value.
print(np.mean(arr))
print(np.std(arr))
print(np.median(arr))
```
