```python
#power function
import numpy as np
arr = np.array([1,2,3,4,5])
print(np.square(arr))
print(np.sqrt(arr))
print(np.power(arr,3))
```

```python
# log Fuctions
arr = np.array([1,2,3,4,5])
print(np.log(arr))
print(np.log10(arr))
print(np.log2(arr))
print(np.exp(arr)) #for exponential values
```

```python
# Round off 
print(np.round(3.13))
print(np.round(3.96)) # for the nearest one digit..
# for round up. we use the ceil function..
print(np.ceil(3.16)) #-->4.0
# for round down .. we use floor funtion.
print(np.floor(3.96)) #-->3.0
# to remove the decimal value we use truncate..
print(np.trunc(3.56)) #-->3.0
```

```python
# Unique values
arr = np.array([1,2,3,4,5,1,2,-1,-5,-8])
print(np.unique(arr))# it gives the unique values..
print(np.sort(arr)) # for sorting..
print(np.abs(arr)) # it converts the negative value to positive and positive remains the same..
```
