```python
#Copy and View
import numpy as np
nums = [1,2,3,4,5]
sub_list = nums[1:4]
print(sub_list)
sub_list[0] = 200
print(sub_list)
print(nums) # for list
#here, copy is created using slicing for list
```

```python
# for Numpy array..

arr = np.array([1,2,3,4,5])
sub_arr = arr[1:4]
print(sub_arr)
sub_arr[0] = 200
print(sub_arr)
print(arr) # here, view is created for the array in slicing
```
